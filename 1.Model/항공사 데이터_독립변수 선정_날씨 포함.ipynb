```python
import pandas as pd

# =========================
# 데이터 로드
# =========================
df = pd.read_csv("new_flight_weather_merged.csv")

target_col = "지연_분"

# ✅ 6개 수치형 변수 전부 포함
num_cols = [
    "기온(°C)",
    "풍속_ms",
    "dep_hour",      # ✅ 다시 포함
    "dep_minute",
    "dep_weekday",
    "is_weekend"
]

# 실제 존재 컬럼만 사용 (안전장치)
num_cols = [c for c in num_cols if c in df.columns]

print("✅ 사용 수치형 컬럼:", num_cols)

# =========================
# 상관관계 계산
# =========================
df_corr = df[num_cols + [target_col]].dropna()

numeric_corr_table = (
    df_corr
    .corr(method="pearson")
    .loc[num_cols, [target_col]]   # ✅ 지연_분 행 제거
    .rename(columns={target_col: "지연_분과의 상관계수"})
    .sort_values(by="지연_분과의 상관계수", ascending=False)
)

numeric_corr_table
```
